```python
# Example input data
input_data = {
    'CreditScore': 600,
    'Geography': 'France',
    'Gender': 'Male',
    'Age': 40,
    'Tenure': 3,
    'Balance': 6000,
    'NumOfProducts': 2,
    'HasCrCard': 1,
    'IsActiveMember': 1,
    'EstimatedSalary': 50000
}
```

```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import pandas as pd
import numpy as np
```

```python
model=load_model("model.h5")
```

```python
## load our sources.
import pickle

with open("label_encoder_gender.pkl","rb") as file:
    label_encoder_gender=pickle.load(file)
with open("onehot_encoder_geo.pkl","rb") as file:
    onehot_encoder_geo=pickle.load(file)
with open("Scaler.pkl","rb") as file:
    Scaler=pickle.load(file)
```

```python
## One-hot_encoded
geo=onehot_encoder_geo.transform([[input_data["Geography"]]])
geo_df=pd.DataFrame(geo.toarray(),columns=onehot_encoder_geo.get_feature_names_out(["Geography"]))
geo_df
```

```python
input_data=pd.DataFrame([input_data])
input_data
```

```python
#Label encoder
input_data["Gender"]=label_encoder_gender.transform([input_data["Gender"]])
input_data
```

```python
data=pd.concat([input_data.drop(columns=["Geography"]),geo_df],axis=1)
data.head(8)
```

```python
data_scaler=Scaler.transform(data)
```

```python
prediction=model.predict(data_scaler)
```

```python
prediction_proba=prediction[0][0]
prediction_proba
```

```python
if prediction_proba > 0.5:
    print("He would like to be here")
else:
    print("He wouldnt")
```
